```python
import pandas as pd
```

```python
data = [
    [8000, 3000, 'default'],
    [2000, 1000, 'default'],
    [   0, 1000, 'default'],
    [5000, 1000, 'ok'],
    [5000, 1000, 'ok'],
    [4000, 1000, 'ok'],
    [9000,  500, 'ok'],
    [3000, 2000, 'default'],
]

df_example = pd.DataFrame(data, columns=['assets', 'debt','status'])
```

```python
from IPython.display import display
```

```python
thresholds = {
    'assets' :[0, 2000, 3000, 4000, 5000, 8000],
    'debt': [500,1000,2000]
}
result = {
    'assets':[],
    'debt': []
}
for feature, Ts in thresholds.items():
    for T in Ts:
        df_left = df_example[df_example[feature] <= T]
        df_right = df_example[df_example[feature] > T]
        left_statuses = df_left.status.value_counts(normalize=True)
        decision_left = left_statuses.idxmax()
        impurity_left = 1 - left_statuses.max()
        right_statuses = df_right.status.value_counts(normalize=True)
        decision_right = right_statuses.idxmax()
        impurity_right = 1 - right_statuses.max()
        avg = float(round((impurity_left + impurity_right)/2,2))
        result[feature].append(avg)
print(result)
```
