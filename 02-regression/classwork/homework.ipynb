```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
url = 'https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv'
df = pd.read_csv(url)
df.head(2)
```

```python
sns.histplot(df.fuel_efficiency_mpg, bins=50)
```

```python

features = ['engine_displacement','horsepower','vehicle_weight','model_year']
target = ['fuel_efficiency_mpg']
df = df[features+target]
df.head()
```

```python
df.isnull().sum()
```

```python
df.horsepower.describe()
```

```python
n = len(df)

n_val = int(n*0.2)
n_test = int(n*0.2)
n_train = n - n_val - n_test

seed = 2
np.random.seed(seed)

idx = np.arange(n)
np.random.shuffle(idx)

df_train = df.iloc[idx[:n_train]]
df_val = df.iloc[idx[n_train:n_train+n_val]]
df_test = df.iloc[idx[n_train+n_val:]]

y_train = df_train.fuel_efficiency_mpg
y_val = df_val.fuel_efficiency_mpg
y_test = df_test.fuel_efficiency_mpg
```

```python
def prepare_X(df, fillWith):
    df = df.copy()
    df_num = df[features].fillna(fillWith)
    X = df_num.values
    return X

def linear(X,y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    return w_full[0], w_full[1:]

def rmse(y, y_pred):
    err = (y-y_pred)**2
    mse = err.mean()
    return np.sqrt(mse)
```

```python
train_hp_mean = df_train.horsepower.mean()
X_train = prepare_X(df_train,train_hp_mean)
w0, w = linear(X_train, y_train)
y_pred = w0 + X_train.dot(w)
train_score = round(rmse(y_train, y_pred),3)
X_val = prepare_X(df_val,train_hp_mean)
y_pred = w0 + X_val.dot(w)
val_score = round(rmse(y_val, y_pred),3)
print('val rmse with mean: ', val_score)
```

```python
X_train = prepare_X(df_train,0)
w0, w = linear(X_train, y_train)
y_pred = w0 + X_train.dot(w)
train_score = round(rmse(y_train, y_pred),3)
X_val = prepare_X(df_val,0)
y_pred = w0 + X_val.dot(w)
val_score = round(rmse(y_val, y_pred),3)
print('val rmse with 0: ', val_score)
```

```python
def linear_reg(X,y,r):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX = XTX + r*np.eye(XTX.shape[0])
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    return w_full[0], w_full[1:]

r_values = [0, 0.01, 0.1, 1, 5, 10,100]
X_train = prepare_X(df_train,0)
results = []
for r in r_values:
    w0, w = linear_reg(X_train, y_train, r)
    X_val = prepare_X(df_val, 0)
    y_pred = w0 + X_val.dot(w)

    score = round(rmse(y_val, y_pred), 4)
    results.append((r,score))

    print('%s: %s' % (r,score))

best_r, best_score = min(results, key=lambda x: (x[1],x[0]))
print(f'Best r: {best_r}')
print(f'Best rmse: {best_score}')
```
