```python
filtered = data[data.origin=='Asia'][['vehicle_weight', 'model_year']].head(7)
X = filtered.to_numpy()
X
```

```python
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
w = XTX_inv.dot(X.T).dot(y)
w.sum()
```
